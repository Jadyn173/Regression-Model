```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import statsmodels.api as sm

pd.options.display.float_format = '{:.2f}'.format
HS_data = pd.read_csv('AICPA_regressionAnalysisData.csv')
display(HS_data)
```

```python
dt4training = HS_data.loc[HS_data['type']=='dt4training']
dt4testing = HS_data.loc[HS_data['type']=='dt4testing']
display(dt4training)
display(dt4testing)
```

```python
y = dt4training.loc[:,'revenue']
x = dt4training.loc[:,['production','coolDD']]

x = sm.add_constant(x)
model1 = sm.OLS(y, x).fit()
model1.params
```

```python
x_test = dt4testing.loc[:,['production', 'coolDD']]
x_test = sm.add_constant(x_test)

model1_assesment = dt4testing.copy()
model1_assesment ['model1_prediction_revenue'] = model1.predict(x_test)
model1_assesment ['model1_abs_pct_error'] = abs(
    (
        model1_assesment['revenue'] - model1_assesment ['model1_prediction_revenue'])/model1_assesment['revenue']
)
display(model1_assesment)
```

```python
model1_assesment['model1_abs_pct_error'].mean()
```

```python
plt.figure(figsize=(10,6))

plt.title('revenue prediction production + coolDD')
plt.ylabel('revenue')
plt.xlabel('date')

plt.plot(model1_assesment['date'],model1_assesment['revenue'], marker='o', color='green')
plt.plot(model1_assesment['date'],model1_assesment['model1_prediction_revenue'], marker='o', color='blue')
```

```python
# ===== UNIVARIATE MODELS (one predictor each) =====

# --- production only ---
y = dt4training.loc[:,'revenue']
x = dt4training.loc[:,['production']]
x = sm.add_constant(x)
model_prod = sm.OLS(y, x).fit()
x_test = dt4testing.loc[:,['production']]
x_test = sm.add_constant(x_test)
pred = model_prod.predict(x_test)
mape = abs((dt4testing['revenue'] - pred) / dt4testing['revenue']).mean()
print('production only            MAPE =', mape)

# --- coolDD only ---
y = dt4training.loc[:,'revenue']
x = dt4training.loc[:,['coolDD']]
x = sm.add_constant(x)
model_cool = sm.OLS(y, x).fit()
x_test = dt4testing.loc[:,['coolDD']]
x_test = sm.add_constant(x_test)
pred = model_cool.predict(x_test)
mape = abs((dt4testing['revenue'] - pred) / dt4testing['revenue']).mean()
print('coolDD only                MAPE =', mape)

# --- heatDD only ---
y = dt4training.loc[:,'revenue']
x = dt4training.loc[:,['heatDD']]
x = sm.add_constant(x)
model_heat = sm.OLS(y, x).fit()
x_test = dt4testing.loc[:,['heatDD']]
x_test = sm.add_constant(x_test)
pred = model_heat.predict(x_test)
mape = abs((dt4testing['revenue'] - pred) / dt4testing['revenue']).mean()
print('heatDD only                MAPE =', mape)


# ===== MULTIPLE REGRESSION MODELS (two or more predictors) =====

# --- production + coolDD ---
y = dt4training.loc[:,'revenue']
x = dt4training.loc[:,['production','coolDD']]
x = sm.add_constant(x)
model_pc = sm.OLS(y, x).fit()
x_test = dt4testing.loc[:,['production','coolDD']]
x_test = sm.add_constant(x_test)
pred = model_pc.predict(x_test)
mape = abs((dt4testing['revenue'] - pred) / dt4testing['revenue']).mean()
print('production + coolDD        MAPE =', mape)

# --- production + heatDD ---
y = dt4training.loc[:,'revenue']
x = dt4training.loc[:,['production','heatDD']]
x = sm.add_constant(x)
model_ph = sm.OLS(y, x).fit()
x_test = dt4testing.loc[:,['production','heatDD']]
x_test = sm.add_constant(x_test)
pred = model_ph.predict(x_test)
mape = abs((dt4testing['revenue'] - pred) / dt4testing['revenue']).mean()
print('production + heatDD        MAPE =', mape)

# --- coolDD + heatDD ---
y = dt4training.loc[:,'revenue']
x = dt4training.loc[:,['coolDD','heatDD']]
x = sm.add_constant(x)
model_ch = sm.OLS(y, x).fit()
x_test = dt4testing.loc[:,['coolDD','heatDD']]
x_test = sm.add_constant(x_test)
pred = model_ch.predict(x_test)
mape = abs((dt4testing['revenue'] - pred) / dt4testing['revenue']).mean()
print('coolDD + heatDD            MAPE =', mape)

# --- production + coolDD + heatDD ---
y = dt4training.loc[:,'revenue']
x = dt4training.loc[:,['production','coolDD','heatDD']]
x = sm.add_constant(x)
model_all = sm.OLS(y, x).fit()
x_test = dt4testing.loc[:,['production','coolDD','heatDD']]
x_test = sm.add_constant(x_test)
pred = model_all.predict(x_test)
mape = abs((dt4testing['revenue'] - pred) / dt4testing['revenue']).mean()
print('production + coolDD + heatDD MAPE =', mape)
```

Therefore heatDD + production is the best because it has the lowest MAPE
